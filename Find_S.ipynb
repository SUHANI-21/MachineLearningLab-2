```python
# ============================================================
# FIND-S ALGORITHM
# Find the most specific hypothesis consistent with
# all positive examples
# ============================================================

import pandas as pd

# ============================================================
# 1. CREATE A SMALL DATASET
# ============================================================

data = {
    'Sky':     ['Sunny', 'Sunny', 'Rainy', 'Sunny', 'Sunny', 'Rainy'],
    'AirTemp': ['Warm',  'Warm',  'Cold',  'Warm',  'Cold',  'Cold'],
    'Humidity':['Normal','High',  'High',  'High',  'Normal','Normal'],
    'Wind':    ['Weak',  'Strong', 'Strong','Weak',  'Weak',  'Strong'],
    'EnjoySport':['Yes', 'Yes', 'No',  'Yes',  'Yes',  'No']
}

df = pd.DataFrame(data)

print("Dataset:")
display(df)
```

```python
# ============================================================
# 2. FIND-S ALGORITHM
# ============================================================

# Get feature columns
features = list(df.columns[:-1])

# Initialize hypothesis with the most specific values
hypothesis = ['ϕ'] * len(features)

print("\nInitial Hypothesis:")
print(hypothesis)

# Process each training example
for index, row in df.iterrows():

    # Consider only positive examples
    if row['EnjoySport'] == 'Yes':

        print(f"\nProcessing positive example {index + 1}:")
        print(row[features].values)

        # First positive example
        if hypothesis == ['ϕ'] * len(features):

            hypothesis = list(row[features])

        else:

            # Compare each attribute
            for i in range(len(features)):

                if hypothesis[i] != row[features[i]]:
                    hypothesis[i] = '?'

        print("Current Hypothesis:")
        print(hypothesis)
```

```python
# ============================================================
# 3. FINAL HYPOTHESIS
# ============================================================

print("\n===================================")
print("FINAL HYPOTHESIS")
print("===================================")

for feature, value in zip(features, hypothesis):
    print(f"{feature} = {value}")
```
